```python
from pathlib import Path

current_dir = Path.cwd().parent

# Move up 1 level (parent) and into a specific folder
target_dir = current_dir / "data"

diary = target_dir / "ce-pumd-interview-diary-dictionary.xlsx"
```

```python
import pandas as pd

diary_excel = pd.ExcelFile(diary)

for sheet in diary_excel.sheet_names:
    df = diary_excel.parse(sheet)
    df.to_csv(f"{sheet}.csv", index=False)
```
